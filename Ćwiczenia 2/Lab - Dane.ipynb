## Brakujące wartości
Brakujące dane w pandas zapisywane są jako `NaN`.
 - Wykrywanie brakujących:`df.info()`, `df.isna()`.
 - Usuwanie (wiersze lub kolumny): `df.dropna(axis=0)`, `df.dropna(axis=1)`.
 - Zastępowanie podaną wartością: `df.fillna()`.
### Przykłady

```python
import pandas as pd
housing = pd.read_csv("housing.csv")
housing.info()
```

```python
housing_na_rows = housing.isna().any(axis=1)
housing[housing_na_rows]
```

```python
housing1 = housing.dropna(axis=0)
housing1.info()
```

```python
housing2 = housing.dropna(axis=1)
housing2.info()
```

```python
housing3 = housing.fillna({"total_bedrooms": housing["total_bedrooms"].mean()})
housing3[housing_na_rows]
```

## Zadanie 3
Wczytuję dane z na_values='?', żeby znaki ? były traktowane jako braki danych. skipinitialspace=True usuwa spacje po przecinkach.

```python
df = pd.read_csv('censusData-with3errors', skipinitialspace=True, na_values='?')

braki = df.isna().sum(axis=1)
display(braki.value_counts().sort_index())

# Usuwam wiersze z największą liczbą braków.
do_usuniecia = df[braki == braki.max()]
display(do_usuniecia)
df = df.drop(index=do_usuniecia.index)
print('Usunięto wierszy:', len(do_usuniecia))
```

```python
procent_brakow = df.isna().mean() * 100
display(procent_brakow.round(2))

# Jeśli brakuje mniej niż 3% wartości, wpisuję najczęstszą wartość.
for kolumna in df.columns:
    if 0 < procent_brakow[kolumna] < 3:
        df[kolumna] = df[kolumna].fillna(df[kolumna].mode()[0])

# Usuwam kolumny, w których nadal są braki.
df = df.dropna(axis=1)
print('Rozmiar danych:', df.shape)
print('Liczba braków:', df.isna().sum().sum())
```
